```python
import json 

def data(arch):
        with open(arch, 'r', encoding='utf-8') as archivo:
            datos = json.load(archivo)
            return datos['ventas']
        return []
```

```python
producto_top = None
total_top = 0
ventas = data('data.json') 
for venta in ventas:
    total = venta['precio_unitario'] * venta['cantidad']
    if total > total_top:
        total_top = total
        producto_top = venta['producto']

print("El producto con más ingresos es", producto_top, total_top)
```

```python
def filtrar_por_vendedor(ventas, nombre_vendedor):
    resultados = []
    for venta in ventas:
        if venta['vendedor'] == nombre_vendedor:
            resultados.append({
                'producto': venta['producto'],
                'vendedor': venta['vendedor']
            })
            
    return resultados
```

```python
vendedor_buscado = "Ana" 
ventas_filtradas = filtrar_por_vendedor(ventas, vendedor_buscado)
for v in ventas_filtradas:
    print("Producto:", v['producto'], "Vendedor:", v['vendedor'])
```

```python
def total_facturado(ventas):
    total_facturado = 0
    for venta in ventas:
        total_facturado += venta['cantidad'] * venta['precio_unitario']
    return total_facturado
def cantidad_vendida(ventas):
    total_cantidad = 0
    for venta in ventas:
        total_cantidad += venta['cantidad']
    return total_cantidad
print("ticket promedio:", total_facturado(ventas) / cantidad_vendida(ventas))
```
